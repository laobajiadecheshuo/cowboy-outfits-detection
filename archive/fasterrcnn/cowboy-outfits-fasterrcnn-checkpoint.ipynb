# Cowboy Outfits 检测 —— PyTorch 版重写

把 Kaggle 原始 starter-kit（MXNet / AutoGluon 的 SSD）按**相同的思路**重写成纯 PyTorch
（torchvision 官方 Faster R-CNN v2，ResNet50-FPN，加载 COCO 预训练权重做迁移学习），
可以直接在你本机的 **myenv**（Python 3.11 + torch 2.13 cu132 + RTX 5060）里跑通，并且**不修改 myenv**。

两点关键适配：

1. **numpy 注入（不装、不改环境）**：myenv 的 `site-packages/numpy` 编译文件损坏（`_core` 下 `.pyd`
   缺失），但按你的要求不重装环境。这里在项目 `vendor/` 下放了一份完好的 numpy 1.26.4，
   在任何 numpy 被 import 之前用 `sys.path.insert(0, ...)` 让它优先加载。想恢复原状时删掉 `vendor/` 即可。
2. **数据路径本地化**：原 notebook 里的 `/kaggle/input/...` 路径全部改成项目本地路径。

流程与原 notebook 一一对应：读 COCO `train.json` → 按类别分层切 train/valid → 训练 →
在 `valid.csv`（决赛换 `test.csv`）上预测 → 生成 `answer.json` + `sample_answer.zip`。

**2026-09-05 优化**：针对「AP_small≈0.02、belt 仅 21 张图、mAP 0.33 见顶回落」三个短板，
做了 6 处改动（标了【改动】注释）：① 输入分辨率 512→640/1024 ② 换 v2 预训练权重
③ 随机水平翻转增强 ④ 余弦退火 + 12 epoch ⑤ belt 过采样 + 扩大验证集 ⑥ TTA + score_thresh 0.05。

```python
# ============================================================
# 环境初始化：路径定位 + vendor numpy 注入 + CUDA 探测 + 导入
# ============================================================
import os, sys, json, zipfile, random, logging
from collections import Counter

# ---- 定位项目根目录（notebook 就在数据文件夹里）----
# 正常情况下 cwd 就是项目根目录；但如果在上一级目录启动 jupyter，
# 就回退到同级的 'CowBoy Outfits Detection' 文件夹，保证后续相对路径都对。
PROJECT_ROOT = os.getcwd()
if not os.path.exists(os.path.join(PROJECT_ROOT, 'train.json')):
    alt = os.path.join(os.path.dirname(PROJECT_ROOT), 'CowBoy Outfits Detection')
    if os.path.exists(os.path.join(alt, 'train.json')):
        PROJECT_ROOT = alt
os.chdir(PROJECT_ROOT)  # 切到项目根目录，后面所有相对路径都以它为基准

# ---- vendor numpy 注入：必须在任何 numpy import 之前 ----
# myenv 自带的 numpy 编译文件损坏（_core 下的 .pyd 缺失），直接 import 会崩。
# 不重装环境，而是把一份完好的 numpy 1.26.4 放在项目 vendor/ 目录，
# 用 sys.path.insert(0, ...) 让它在 sys.path 中优先级最高，从而抢先被加载。
# 注意：这行必须出现在任何 import numpy（包括被 torch/pandas 间接导入）之前。
VENDOR = os.path.join(PROJECT_ROOT, 'vendor')
if os.path.isdir(os.path.join(VENDOR, 'numpy')):
    sys.path.insert(0, VENDOR)

import numpy as np
import torch

# CUDA 探测必须在 import torchvision 之前做：
# 这台机器上若先加载 torchvision 的 CUDA ops 再调用 torch.cuda.is_available() 会段错误，
# 所以先在这里探测并固定 device，之后才 import torchvision。
HAS_CUDA = torch.cuda.is_available()
device = torch.device('cuda' if HAS_CUDA else 'cpu')
GPU_NAME = torch.cuda.get_device_name(0) if HAS_CUDA else ''

import torchvision
import torchvision.transforms.functional as F  # F.to_tensor: PIL图像 -> [0,1] 浮点张量
from torchvision.models.detection import fasterrcnn_resnet50_fpn_v2, FasterRCNN_ResNet50_FPN_V2_Weights
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor  # 用于替换检测头做迁移学习
from pycocotools.coco import COCO          # 解析 COCO 格式标注文件 train.json
from pycocotools.cocoeval import COCOeval  # 官方 mAP 评估器
import pandas as pd
from PIL import Image

# 防御性检查：确认 numpy 确实是从 vendor/ 加载的，否则说明注入失败
assert 'vendor' in np.__file__, 'numpy 应来自 vendor/，检查 sys.path 注入'

# 固定随机种子，保证划分数据集、训练过程可复现
torch.manual_seed(0); np.random.seed(0); random.seed(0)

# 配置日志输出到 stderr（pycocotools 等库会往 logger 里写评估信息）
logger = logging.getLogger()
logger.addHandler(logging.StreamHandler(sys.stderr))

# 冒烟测试开关：设环境变量 COWBOY_QUICK=1 后用小数据 + 少量 epoch 快速验证流程，
# 正常训练不用管这个变量。
QUICK = os.environ.get('COWBOY_QUICK') == '1'

# 打印环境信息，便于排查“跑错环境/走错设备”的问题
print('numpy   ', np.__version__, '@', os.path.dirname(np.__file__))
print('torch   ', torch.__version__, '| cuda:', HAS_CUDA, GPU_NAME)
print('torchvision', torchvision.__version__)
print('device  ', device)
print('root    ', PROJECT_ROOT)
```

## Load the training data

```python
# ============================================================
# 读取 COCO 格式标注 train.json，建立类别映射
# ============================================================
# COCO 是 pycocotools 提供的解析器，能方便地按图片/类别查询标注
coco = COCO(os.path.join(PROJECT_ROOT, 'train.json'))
print('Data info:', coco.dataset.get('info'))

# 构建 {类别名: COCO类别id} 字典，本数据集共 5 类牛仔服饰
# （如 belt / boots / cowboy_hat / jacket / sunglasses）
categories = {cat['name']: cat['id'] for cat in coco.loadCats(coco.getCatIds())}
print('Categories:', categories)

# ---- 类别编号的双向映射 ----
# 关键点：COCO 的类别 id 是任意整数（不一定从 1 连续），
# 而 Faster R-CNN 要求训练标签是 1..N 的连续整数（0 保留给背景类）。
# 所以建立两套映射，训练时用训练编号，输出结果时再换回 COCO id。
class_names = list(categories.keys())                                          # 顺序即训练类别编号 1..5
cat_id_to_label = {categories[n]: i + 1 for i, n in enumerate(class_names)}    # COCO id -> 训练编号(1~5)
cat_label_to_id = {i + 1: categories[n] for i, n in enumerate(class_names)}    # 训练编号(1~5) -> COCO id
print('class_names:', class_names)
print('cat_label_to_id:', cat_label_to_id)
```

```python
# ============================================================
# 把所有标注整理成 DataFrame：每张图片一行，rois 列存该图所有目标框
# ============================================================
def load_rois(coco, img):
    """取出一张图片的全部标注框，转换成归一化坐标（0~1）的字典列表。

    COCO 的 bbox 格式是 [x, y, w, h]（左上角坐标 + 宽高，单位像素），
    这里转成 [xmin, ymin, xmax, ymax] 且除以图像宽高做归一化，
    与原 MXNet notebook 的数据表示保持一致。
    """
    anns = coco.loadAnns(coco.getAnnIds(imgIds=img['id']))  # 按图片 id 查它所有标注
    rois = []
    for a in anns:
        x, y, w, h = a['bbox']
        rois.append({'class': next(n for n, cid in categories.items() if cid == a['category_id']),
                     'xmin': x / img['width'],  'ymin': y / img['height'],
                     'xmax': (x + w) / img['width'], 'ymax': (y + h) / img['height'],
                     'area': a['area'],                 # 框面积（像素），COCO 评估区分大中小目标时用
                     'iscrowd': a.get('iscrowd', 0)})   # 是否为密集人群标注（训练时跳过）
    return rois


# 逐图构建记录：图片绝对路径、id、宽高、目标框列表
rows = [{'image': os.path.join(PROJECT_ROOT, 'images', img['file_name']),
         'image_id': img['id'], 'width': img['width'], 'height': img['height'],
         'rois': load_rois(coco, img)} for img in coco.dataset['images']]
train = pd.DataFrame(rows)
print('total images:', len(train))

# 统计各类别的标注框总数，看类别是否均衡
cnt = Counter()
for rois in train['rois']:
    for r in rois:
        cnt[r['class']] += 1
print('class distribution:', dict(cnt))

# ---- 按类别分层切分 train / valid ----
# 每个类别随机抽若干张“包含该类别”的图进验证集，保证验证集中每个类别都有样本
# （一张图可能含多类，所以要去重）。
# 【改动】验证集从每类 10 张扩到最多 20 张：原来只有 50 张验证图，mAP 噪声 ±0.02 起步，
# epoch 之间的波动很可能只是噪声，调参结论不可靠。
# 但稀有类别要封顶：belt 全数据集只有 21 张图，若也抽 20 张，训练集里 belt 就一张不剩了。
# 所以每类最多抽「该类图片数的 1/5」（下限 2 张、上限 20 张）。
sample_n_per_cat = 20
valid_ids = []
for cat_name in categories:
    # 找出所有包含该类别的图片
    mask = train['rois'].apply(lambda rois: any(r['class'] == cat_name for r in rois))
    df = train[mask]
    n_valid = min(sample_n_per_cat, max(2, len(df) // 5), len(df))  # 稀有类少留验证、多留训练
    df = df.sample(n_valid, random_state=0)
    valid_ids += list(df.index)
valid_ids = list(set(valid_ids))  # 一张图被多个类别抽中时只保留一次
train_ids = [i for i in train.index if i not in valid_ids]
train_data = train.loc[train_ids]
valid_data = train.loc[valid_ids]
print('train split:', len(train_data), 'valid split:', len(valid_data))

# ---- 【改动】belt 过采样：belt 全数据集只有 21 张图 / 25 个框，模型几乎学不到 ----
# 把含 belt 的训练图复制几份加进训练集。配合 Dataset 里的随机水平翻转，
# 同一张图每个 epoch 看到的增强不同，重复不等于死记硬背。
BELT_COPIES = 6  # 过采样后 belt 图的总份数（1 份原始 + 5 份重复）
belt_mask = train_data['rois'].apply(lambda rois: any(r['class'] == 'belt' for r in rois))
print('belt images in train:', int(belt_mask.sum()))
if belt_mask.sum() > 0:
    train_data = pd.concat([train_data] + [train_data[belt_mask]] * (BELT_COPIES - 1),
                           ignore_index=True)
    print(f'belt oversample x{BELT_COPIES} -> train split: {len(train_data)}')

if QUICK:  # 冒烟测试：压缩数据量，只验证流程能跑通
    train_data = train_data.sample(min(64, len(train_data)), random_state=0)
    valid_data = valid_data.sample(min(20, len(valid_data)), random_state=0)
    print('quick mode -> train:', len(train_data), 'valid:', len(valid_data))
```

```python
# ============================================================
# PyTorch Dataset：把 DataFrame 包装成 torchvision 检测模型要求的格式
# ============================================================
# 【重要】Dataset 类和 collate_fn 定义在同目录的 cowboy_dataset.py 里，
# 而不是直接写在本 cell 中。原因：Windows 下 num_workers>0 用 spawn 创建子进程，
# 子进程是全新解释器，必须能 import Dataset 类所在的模块；
# notebook 里的类属于 __main__，子进程 import 不到 Jupyter 的 __main__，
# worker 会静默死掉、主进程永远等不到数据（表现为“运行没反应”）。
# 定义在 .py 模块里后，多进程数据加载才能正常工作。
from cowboy_dataset import COCODetectionDataset, collate_fn

# categories / cat_id_to_label 要显式传入构造函数：
# worker 子进程里没有 notebook 的全局命名空间，Dataset 必须自带这两张映射表
# 【改动】训练集开 hflip=True（随机水平翻转增强，见 cowboy_dataset.py）；
# 验证集保持原图不增强，评估结果才可复现
train_ds = COCODetectionDataset(train_data, categories, cat_id_to_label, hflip=True)
valid_ds = COCODetectionDataset(valid_data, categories, cat_id_to_label)
print('train_ds:', len(train_ds), '| valid_ds:', len(valid_ds))
```

## 训练（Faster R-CNN v2，迁移 COCO 预训练权重）

首次运行会从 download.pytorch.org 下载 `fasterrcnn_resnet50_fpn_v2` 的 COCO 预训练权重（约 167MB）。

```python
# ============================================================
# 构建模型：Faster R-CNN v2 (ResNet50-FPN V2) + COCO 预训练权重，替换检测头
# ============================================================
num_classes = len(class_names) + 1  # 5 个服饰类 + 1 个背景类（背景必须是第 0 类）

# 【改动 1】换用 v2 预训练模型：v2 在 COCO 80 类上 mAP 46.7，比 v1 的 37.0 高出一大截
# （RPN/检测头结构、训练配方都更强），对小目标也明显更友好。
# 首次运行会从 download.pytorch.org 下载约 167MB 权重。
# 【改动 2】输入分辨率 512→640/1024：旧配置下 AP_small≈0.02、小目标几乎全军覆没，
# 分辨率是小目标的第一瓶颈（belt 在 512 短边下可能只剩几十个像素）。
# 计算量随边长平方增长，显存不够时：先降 batch（见训练 cell），再降 min_size 到 560/512。
model = fasterrcnn_resnet50_fpn_v2(weights=FasterRCNN_ResNet50_FPN_V2_Weights.COCO_V1,
                                   min_size=640, max_size=1024)

# 迁移学习关键一步：预训练模型的分类/回归头是 91 类（COCO），
# 把它换成我们任务的 6 类（5 类 + 背景）。
# v2 的 cls_score.in_features 同样是 1024，替换方式与 v1 完全一致。
in_features = model.roi_heads.box_predictor.cls_score.in_features
model.roi_heads.box_predictor = FastRCNNPredictor(in_features, num_classes)

# device 已在第一个 cell 里定好（必须在 import torchvision 之前探测 CUDA）
model.to(device)
print('model on', device, '| num_classes =', num_classes)
```

```python
# ============================================================
# 验证集推理 + COCO 官方 mAP 评估（含 TTA 测试时增强 + 按类别 AP 诊断）
# ============================================================
def _merge_tta(pred, pred_flip, img_w, iou_thresh=0.5):
    """合并「原图预测」和「水平翻转图预测」两份结果（TTA 测试时增强）。

    翻转图上的框先映射回原图坐标系（x' = W - x），再和原图结果拼在一起，
    用按类别分组的 NMS（batched_nms，不同类别互不抑制）去掉两组互相重复的框。
    """
    boxes_f = pred_flip['boxes'].clone()
    boxes_f[:, [0, 2]] = img_w - boxes_f[:, [2, 0]]   # 翻转回原图坐标
    boxes = torch.cat([pred['boxes'], boxes_f])
    scores = torch.cat([pred['scores'], pred_flip['scores']])
    labels = torch.cat([pred['labels'], pred_flip['labels']])
    keep = torchvision.ops.batched_nms(boxes, scores, labels, iou_thresh)
    return {'boxes': boxes[keep], 'scores': scores[keep], 'labels': labels[keep]}


def predict_batch(model, loader, device, tta=True):
    """在整个 DataLoader 上推理，返回 COCO 结果格式的字典列表：
    [{image_id, category_id, bbox:[x,y,w,h], score}, ...]
    tta=True 时每张图额外推理一次水平翻转版，两份结果合并（见 _merge_tta）。
    对 AP 指标通常能白拿零点几个点，代价是推理时间翻倍。
    """
    model.eval()
    results = []
    with torch.no_grad(), torch.autocast(device_type='cuda', dtype=torch.bfloat16, enabled=HAS_CUDA):
        # 【提速改动】autocast(bf16)：和训练侧一样加速推理；
        # 这里直接用 HAS_CUDA 判断，保证本 cell 可独立运行
        for images, targets in loader:
            images = [im.to(device, non_blocking=True) for im in images]
            preds = model(images)  # eval 模式下模型直接输出预测框/类别/分数
            if tta:
                flipped = [torch.flip(im, dims=[2]) for im in images]  # 沿 W 维翻转
                preds_f = model(flipped)
                preds = [_merge_tta(p, pf, im.shape[2])
                         for p, pf, im in zip(preds, preds_f, images)]
            for t, p in zip(targets, preds):
                img_id = int(t['image_id'])
                for box, label, score in zip(p['boxes'], p['labels'], p['scores']):
                    x1, y1, x2, y2 = box.tolist()
                    results.append({'image_id': img_id,
                                    # 训练编号换回原始 COCO 类别 id，才能和 GT 对齐评估
                                    'category_id': cat_label_to_id[int(label)],
                                    # COCO 评估要求的 bbox 格式是 [x, y, w, h]
                                    'bbox': [x1, y1, x2 - x1, y2 - y1],
                                    'score': float(score)})
    return results


def evaluate_map(model, loader, device, verbose=True):
    """用 pycocotools 的 COCOeval 计算 mAP（IoU=0.5:0.95，即 COCO 主指标）。"""
    results = predict_batch(model, loader, device)
    gt_ids = list(set(valid_data['image_id']))

    # 【重要坑】本数据集的 image id 是 uint64，可能超过 int64 上限；
    # COCOeval 内部用 numpy 处理 id 时会静默转成 float64 丢失精度，
    # 导致 GT 和检测结果匹配不上、mAP 恒为 -1。
    # 解决办法：先把所有 id 映射成 0..N-1 的连续小整数，GT 和 DT 用同一套映射。
    id2small = {uid: i for i, uid in enumerate(gt_ids)}

    # 从原始 COCO 数据里抽出验证集对应的 images / annotations，并替换为小整数 id
    gt = {'images': [{'id': id2small[im['id']],
                      'width': im['width'], 'height': im['height'], 'file_name': im['file_name']}
                     for im in coco.dataset['images'] if im['id'] in id2small],
          'annotations': [dict(a, image_id=id2small[a['image_id']])
                          for a in coco.dataset['annotations'] if a['image_id'] in id2small],
          'categories': coco.dataset['categories']}
    # 检测结果同样映射成小整数 id
    results = [dict(r, image_id=id2small[r['image_id']]) for r in results]

    # 本机 pycocotools 的 COCO() 只接受文件路径（不支持直接传 dict），
    # 所以先把 GT 落盘成临时 json 再读入，用完即删
    import tempfile
    with tempfile.NamedTemporaryFile('w', suffix='.json', delete=False, encoding='utf-8') as f:
        json.dump(gt, f)
        gt_path = f.name
    try:
        gt_coco = COCO(gt_path)
    finally:
        os.remove(gt_path)

    dt_coco = gt_coco.loadRes(results)       # 把检测结果挂到 GT 上
    ev = COCOeval(gt_coco, dt_coco, 'bbox')
    ev.evaluate(); ev.accumulate()
    if verbose:
        # ---- 【改动】按类别打印 AP@IoU=0.50：定位拖后腿的类（如稀有类 belt）----
        # precision 形状 [T,R,K,A,M]：IoU阈值 x 召回点 x 类别 x 面积范围 x maxDets
        prec = ev.eval['precision']
        for k, cid in enumerate(ev.params.catIds):
            p = prec[:, :, k, 0, -1]         # 所有 IoU 阈值、area=all、maxDets=100
            ap50 = p[p > -1].mean() if (p > -1).any() else float('nan')
            print(f"    AP50[{coco.loadCats([cid])[0]['name']:>10s}] = {ap50:.3f}")
        ev.summarize()  # 打印 AP / AP50 / AP75 / 大中小目标 AP 等完整指标
    return ev.stats[0]   # stats[0] 就是 mAP@[.5:.95]
```

```python
# ============================================================
# 训练循环：SGD + 余弦退火，每个 epoch 后在验证集上评估 mAP
# ============================================================
# 【改动】5→12 个 epoch：原来恒定学习率在 epoch 3 就见顶回落（过拟合苗头），
# 配合数据增强 + 学习率衰减，训练更久还能继续涨
EPOCHS = 2 if QUICK else 12
# 【改动】batch 8→4：v2 模型更大 + 输入分辨率更高，显存占用上升；
# RTX 5060 Laptop (8GB) 上 batch 4 比较稳，OOM 就降到 2
BATCH_SIZE = 4

# 只优化需要梯度的参数（本 notebook 没有冻结任何层，即全网络微调）
params = [p for p in model.parameters() if p.requires_grad]
# 检测任务经典 SGD 配置：lr=0.005, momentum=0.9, weight_decay=5e-4
optimizer = torch.optim.SGD(params, lr=0.005, momentum=0.9, weight_decay=0.0005)
# 【改动】余弦退火：学习率从 0.005 平滑衰减到接近 0，
# 末期小学习率用来精修框定位，避免恒定 lr 那种后期 mAP 回落
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)

# 【提速改动】num_workers=4：用子进程并行读图/解码，GPU 不再空等 CPU 准备数据；
# persistent_workers=True：epoch 之间不销毁重建 worker 进程，省去每个 epoch 的启动开销；
# pin_memory=True：数据先放进锁页内存，CPU→GPU 拷贝更快。
# 前提：Dataset 已移到 cowboy_dataset.py（Windows spawn 子进程需要能 import 它），
# 否则会静默卡死；若仍有异常，把 num_workers 降回 0（同时 persistent_workers 改 False）
train_loader = torch.utils.data.DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                                           collate_fn=collate_fn, num_workers=0, drop_last=True,
                                           persistent_workers=False, pin_memory=True)
# 验证集：batch=1 逐张评估，不打乱，保证结果可复现
valid_loader = torch.utils.data.DataLoader(valid_ds, batch_size=1, shuffle=False,
                                           collate_fn=collate_fn, num_workers=0,
                                           persistent_workers=False, pin_memory=True)

# 【提速改动】混合精度（bf16）：RTX 5060 原生支持 bf16 张量核计算，
# 前向/反向大约快 1.5~2 倍，且 bf16 数值范围与 fp32 相同，
# 不需要 fp16 那套 GradScaler 梯度缩放，训练和原 fp32 一样稳定
USE_AMP = HAS_CUDA


def train_one_epoch(model, loader, optimizer, device, epoch):
    """训练一个 epoch，返回平均 loss。

    注意：Faster R-CNN 在 train 模式下不输出预测框，而是返回一个 loss 字典：
    - loss_classifier:  RoiHead 的分类损失
    - loss_box_reg:     RoiHead 的框回归损失
    - loss_objectness:  RPN 的前景/背景二分类损失
    - loss_rpn_box_reg: RPN 的锚框回归损失
    四项相加作为总 loss 反向传播。
    """
    model.train()
    total, nb = 0.0, 0
    for images, targets in loader:
        # non_blocking=True 配合 pin_memory：异步搬运到 GPU，与计算重叠
        images = [im.to(device, non_blocking=True) for im in images]
        # image_id 是 Python int 不是 tensor，不能 .to(device)，用条件表达式跳过
        targets = [{k: (v.to(device) if isinstance(v, torch.Tensor) else v) for k, v in t.items()} for t in targets]
        # autocast 只包住前向：内部自动把卷积/矩阵乘转成 bf16，loss 计算保持 fp32
        with torch.autocast(device_type='cuda', dtype=torch.bfloat16, enabled=USE_AMP):
            loss_dict = model(images, targets)  # train 模式：前向即算 loss
            loss = sum(v for v in loss_dict.values())
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total += loss.item(); nb += 1
        if nb % 50 == 0:  # 每 50 个 batch 打印一次分项 loss，便于观察收敛情况
            detail = ' '.join(f'{k}={v.item():.3f}' for k, v in loss_dict.items())
            print(f'  [epoch {epoch}][batch {nb}] loss={loss.item():.4f} {detail}', flush=True)
    return total / nb


# ---- 主训练循环：每个 epoch 训练完立刻在验证集上算 mAP ----
for epoch in range(EPOCHS):
    avg = train_one_epoch(model, train_loader, optimizer, device, epoch)
    print(f'[Epoch {epoch}] avg train loss: {avg:.4f} | lr: {scheduler.get_last_lr()[0]:.5f}', flush=True)
    mAP = evaluate_map(model, valid_loader, device)
    print(f'[Epoch {epoch}] valid mAP: {mAP:.4f}', flush=True)
    scheduler.step()  # 余弦退火按 epoch 衰减学习率
```

## Generate submission

`valid.csv` 用于公开榜提交，`test.csv` 用于决赛提交。决赛只有 3 次机会，别在最后一天传错文件。


```python
# ============================================================
# 对 valid.csv / test.csv 中的图片逐张推理，生成比赛提交结果
# ============================================================
def create_submission(df, model, device, score_thresh=0.05, tta=True):
    """对 DataFrame 里的每张图推理，返回提交格式的字典列表。

    参数 score_thresh：置信度阈值，低于该分数的框直接丢弃。
    【改动】0.1→0.05：评测按 AP 计分（每图最多取 top-100 检测），
    多保留低分框一般不掉分、反而可能提高召回，压到模型默认输出下限 0.05 即可。
    tta=True：与 evaluate_map 一致，做水平翻转测试时增强（见 _merge_tta），代价是推理时间翻倍。
    """
    results = []
    model.eval()
    with torch.no_grad(), torch.autocast(device_type='cuda', dtype=torch.bfloat16, enabled=HAS_CUDA):
        for _, row in df.iterrows():
            img_id = int(row['id'])
            # 强制转 RGB：测试集里混有 RGBA 四通道图，不转的话模型内部归一化会报错
            img = Image.open(row['file_name']).convert('RGB')
            img_t = F.to_tensor(img).to(device)
            pred = model([img_t])[0]  # 模型接收 list，返回 list；取第 0 个即本图结果
            if tta:
                pred_f = model([torch.flip(img_t, dims=[2])])[0]
                pred = _merge_tta(pred, pred_f, img_t.shape[2])
            for box, label, score in zip(pred['boxes'], pred['labels'], pred['scores']):
                if score.item() > score_thresh:
                    x1, y1, x2, y2 = box.tolist()
                    results.append({'image_id': img_id,
                                    'category_id': cat_label_to_id[int(label)],  # 换回 COCO 类别 id
                                    # 注意：比赛要求的提交格式是 [x1,y1,x2,y2] 像素坐标，
                                    # 与 COCO 评估用的 [x,y,w,h] 不同，别搞混
                                    'bbox': [x1, y1, x2, y2],
                                    'score': float(score)})
    return results
```
